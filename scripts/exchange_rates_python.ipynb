```python
import requests
import pandas as pd

currencies = ['usd', 'eur', 'chf', 'jpy', 'krw']
start_date = '2025-01-01'
end_date = '2026-01-01'

all_dataframes = []

for currency in currencies:
    url = f'https://api.nbp.pl/api/exchangerates/rates/a/{currency}/{start_date}/{end_date}/?format=json'
    response = requests.get(url)
    data = response.json()

    df = pd.DataFrame(data['rates'])
    df['currency'] = currency.upper()
    df['effectiveDate'] = pd.to_datetime(df['effectiveDate'])

    all_dataframes.append(df)

df_all = pd.concat(all_dataframes, ignore_index=True)

print("Datatypes : ")
df_all.info()

print("\n\nRows per currency : ")
print(df_all['currency'].value_counts())

df_all.head()

df_all.to_csv('nbp_rates.csv', index=False)
```
